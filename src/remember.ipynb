# Math in README with LaTeX

You can write inline math using `$...$` or display math using `$$...$$`.

---

### Example: Pythagoras’ Theorem  

Inline: The Pythagoras theorem is written as $a^2 + b^2 = c^2$.  

Block display:  

$$
a^2 + b^2 = c^2
$$
